```python
l = list(map(int, input("Entre numero: ").split()))
d = dict()
for i in l:
    if i not in d.keys():
        d[i] = 1
    else :
        d[i] += 1
mx = -1
for i in d.keys():
    if d[i] > mx:
        mx = i
mx
```
