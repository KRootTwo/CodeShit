6. Write a program to reverse a list and compare the time complexity: (I) using a loop
and (ii) using slicing.

```python
import timeit
# Loop
def rev_w_loop (ls):
    left, right = 0, len(ls) - 1

    while left < right :
        ls[left], ls[right] = ls[right], ls[left]
        left += 1
        right -= 1
    return ls

# Slicing
def rev_w_slice (ls):
    return ls[::-1]

test_list = list(range(10000))

# main
time_of_loop = timeit.timeit(lambda: rev_w_loop(test_list.copy()), number = 1000)
time_of_slce = timeit.timeit(lambda: rev_w_slice(test_list), number = 1000)

print(f"Time in loop {time_of_loop:.5f} seconds")
print(f"Time in slice {time_of_slce:.5f} seconds")
```

7. Write a program to check if a number is prime. Improve the algorithm from O(n) to O(n).

```python
import math
import timeit

# O(n)
def is_prime_n(n):
    for i in range(2, n - 1):
        if n % i == 0:
            return False
    return True

# O(log(n))
def is_prime_logn(n):
    for i in range(2, int(math.sqrt(n))):
        if n % i == 0:
            return False
    return True

a = int(input("Entre numero bludclaat: "))

time_n = timeit.timeit(lambda: is_prime_n(a), number = 1000)
time_logn = timeit.timeit(lambda: is_prime_logn(a), number = 1000)

print(f"time for O(n): {time_n:.5f} sec")
print(f"time for O(log(n)): {time_logn:.5f} sec")
```

11. Write a python program to implement binary search. Compare runtime of       Linear Search vs Binary Search using timeit for:
    • n = 100
    • n = 1000
    • n = 10000
Record observations.

```python
import timeit

def linear_search(ls, tg):
    for i, j in enumerate(ls):
        if j == tg:
            return i
    return -1

def binary_search(ls, tg):
    start, end = 0, len(ls) - 1
    while start <= end:
        mid = start + (end - start) // 2
        if ls[mid] == tg:
            return mid
        elif ls[mid] > tg:
            end = mid - 1
        else:
            start = mid + 1
    return -1

test_list = list(range(10000))

time_l1 = timeit.timeit(lambda: linear_search(test_list, 9999), number=100)
time_b1 = timeit.timeit(lambda: binary_search(test_list, 9999), number=100)

time_l2 = timeit.timeit(lambda: linear_search(test_list, 9999), number=1000)
time_b2 = timeit.timeit(lambda: binary_search(test_list, 9999), number=1000)

time_l3 = timeit.timeit(lambda: linear_search(test_list, 9999), number=10000)
time_b3 = timeit.timeit(lambda: binary_search(test_list, 9999), number=10000)

print(f"Linear time for 100 obs: {time_l1:.5f}")
print(f"Binary time for 100 obs: {time_b1:.5f}\n")
print(f"Linear time for 1000 obs: {time_l2:.5f}")
print(f"Binary time for 1000 obs: {time_b2:.5f}\n")
print(f"Linear time for 10000 obs: {time_l3:.5f}")
print(f"Binary time for 10000 obs: {time_b3:.5f}\n")
```

13. Write a program that generates all subsets of a list. Determine time complexity with explanation.

```python
import itertools
import timeit

# Normal
def get_all_subsets_n(set):
    subs = [[]]

    for i in set:
        items = [e + [i] for e in subs]
        subs.extend(items)
    return subs

test_set = list(range(15))

# Using itertools
def get_all_subsets_i(set):
    subs = []
    for i in range(len(set) + 1):
        for c in itertools.combinations(set, i):
            subs.append(list(c))
    return subs


time_sub_n = timeit.timeit(lambda: get_all_subsets_n(test_set), number=1000)
time_sub_i = timeit.timeit(lambda: get_all_subsets_i(test_set), number=1000)

print(f"{time_sub_n:.5f}")
print(f"{time_sub_i:.5f}")
```

15. Write a recursive Fibonacci function and analyze the time complexity. Compare the time complexity with the time complexity of an iterative Fibonacci function.

```python

import timeit

def fibo_slow(n):
    if n <= 1:
        return n
    return fibo_slow(n - 1) + fibo_slow(n - 2)

def fibo_fast(n):
    if n <= 1:
        return n 
    a, b = 0, 1

    for _ in range(2, n + 1):
        a, b = b, a + b 
    return b

time_fs = timeit.timeit(lambda: fibo_slow(20), number=1000)
time_ff = timeit.timeit(lambda: fibo_fast(20), number=1000)

print(f"{time_fs:.5f}")
print(f"{time_ff:.5f}")
```

16. Write a program to identify the element that appears most frequently in a list. Measure the execution time of the program and discuss its time complexity.
