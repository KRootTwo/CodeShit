17.  Write a Python NumPy program to solve the linear system Ax = b where
                A = |1 2|   B = | 5|
                    |3 4|       |11|
    using numpy.linalg.solve().

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
b = np.array([5, 11])

x = np.linalg.solve(A, b)

print(f"{list(map(int, x))}")
```

18. Write a Python program to compute the determinant of the matrix
                A = |2 1|
                    |6 3| 
    and determine whether the system Ax = b has a unique solution.

```python
import numpy as np

A = np.array([[2, 1], [6, 3]])

det_A = np.linalg.det(A)

print(f"{det_A}")

if np.isclose(det_A, 0):
    print(f"No unique solution")
else:
    print(f"Unique solution")
```
