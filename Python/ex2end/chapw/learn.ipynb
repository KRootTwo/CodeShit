Lambda Expressions

```python
def cube1(x):
    return x**3

# similar

cube2 = lambda x: x**3

print(cube1(2))
print(cube2(2))
```

```python
def fun (fn, x):
    return (fn(x)) // 10

print(fun(lambda x: x ** 2, 98))
```
