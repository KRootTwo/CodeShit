7. Write a Python program to:
    (a) Compute P (X = 25) for 50 fair coin tosses.
    (b) Print probabilities for k = 20 to 30.
    (c) Verify that sum of all probabilities equals 1.

```python
import math

nCkp = lambda n, k, p: (math.factorial(n)/(math.factorial(n-k)*math.factorial(k))) * (p ** k) * (1 - p) ** (n - k)

binomial_pmf = lambda n, k, p: math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

n = 50
p = 0.5

print(f"Binomial PMF for P(x = 25) is {nCkp(n, 25, p):.5f}")

# 2
for i in range (20, 30 + 1):
    print(f"P(X = {i}) = {nCkp(n, i, p):.6f}")

# 3
total_prob = sum(nCkp(n, k, p) for k in range(51))
print(f"{total_prob}")
```

8. 8 bits are transmitted independently with probability of error p = 0.2.
    Write a Python program to:
    (a) Compute P (X = k) for k = 0 to 8.
    (b) Print probability table.
    (c) Plot the PMF.

```python
import math 
import matplotlib.pyplot as plt

bino_pmf = lambda n, k, p: math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

# 1 & 2
n = 8
p = 0.2

k = list(range(0, 9))
probs = []

for i in k:
    probs.append(bino_pmf(n, i, p))
    print(f"P(X = {i}) = {bino_pmf(n, i, p):.5f}")


# 3
plt.bar(k, probs)
plt.xlabel("Number of Corrupted Bits (k)")
plt.ylabel("Probability P(X = k)")
plt.title("Binomial PMF for 8-Bit Transmission (p = 0.2)")
plt.show()
```

9. A die is rolled 200 times with frequencies:
        (20, 35, 40, 45, 30, 30)
    Write a Python program to:
        (a) Construct empirical PMF.
        (b) Compute empirical mean.
        (c) Compute empirical variance.
        (d) Print results clearly.

```python
freqs = [20, 35, 40, 45, 30, 30]
outcomes = [1, 2, 3, 4, 5, 6]
total = sum(freqs)

empirical_pmf = [f/total for f in freqs]
empirical_mean = sum([(x * p) for x, p in zip(outcomes, empirical_pmf)])
empirical_variance = sum(
    ((x - empirical_mean) ** 2) * p for x, p in zip(outcomes, empirical_pmf)
)
print(f"{empirical_pmf}")
print(f"{empirical_mean}")
print(f"{empirical_variance}")
```

10. For a fair die, write a Python program to:
    (a) Compute E[X].
    (b) Compute E[X^2].
    (c) Compute Var(X) using:
        • Definition formula
        • E[X^2] − (E[X])^2
    (d) Verify both results are equal.

```python
import random
from collections import Counter

outcomes = [1, 2, 3, 4, 5, 6]
simualation = [random.randint(1, 6) for _ in range(10000)]
freq = Counter(simualation)
sfreq = dict(sorted(freq.items()))

Px = [(f / 10000) for f in sfreq.values()]
Ex = sum((x * p) for x, p in zip(outcomes, Px))
Var = sum((((x - Ex) ** 2) * p for x, p in zip(outcomes, Px)))
Ex2 = sum(((x ** 2) * p for x, p in zip(outcomes, Px)))
Var2 = Ex2 - (Ex ** 2)


print(f"{Ex}")
print(f"{Var}")
print(f"{Var2}")
```
