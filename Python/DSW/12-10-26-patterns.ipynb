```python
for _ in range(3):
    print('*')
```

```python
print("\n".join("*" for _ in range(3)))
```

```python
for _ in range(3):
    print('*', end = "")
```

```python
n = 4
for i in range(n):
    for j in range(i):
        print('*', end = " ")
    print()
```

```python
for i in range(n):
    print('* ' * (i + 1))
```

```python
for i in range (n):
    print(" " * (i), '*')
```

```python
for i in range (n):
    print(" " * (n - i - 1), '*')
```

```python
for i in range (n):
    print(" " * (n - i - 1), '*' * (i + 1))
```

```python
for i in range (n):
    print(f"{" " * (n - i - 1)}{"*" * (i)}{"*" * (i + 1)}")
```

```python
for i in range (n):
    print(" " * (n - i - 1), "*" * (i * 2 + 1))
```

```python
for i in range (n):
    print(" " * (i), "*" * ((n - i) * 2 - 1))
```

```python
for i in range (n - 1):
    print(" " * (n - i - 1), "*" * (i * 2 + 1))

for i in range (n):
    print(" " * (i), "*" * ((n - i) * 2 - 1))
```
