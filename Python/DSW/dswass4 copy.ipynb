import numpy as np

- Predict all four outputs.

````python
# Line-by-line trace:
# A = [[1, 2, 3], [4, 5, 6]] (shape 2x3)
# B = [6] (shape (1,), view of A[1, 2])
# C = A[1:][2] -> IndexError! A[1:] has shape (1, 3), so index 2 is out of bounds.

# Fixed version (using .copy() so modifying D doesn't mutate A):
A = np.arange(1, 7).reshape(2, 3)
B = A[1:, 2].copy()
D = B.T
D[0] = 100

print("A:\n", A)
print("B:", B)
print("D:", D)
````

**Trace & Fix:**
- `A = np.arange(1, 7).reshape(2, 3)` $\rightarrow$ `[[1, 2, 3], [4, 5, 6]]`
- `B = A[1:, 2]` $\rightarrow$ `[6]` (1D view pointing to `A[1, 2]`)
- `C = A[1:][2]` $\rightarrow$ Throws `IndexError: index 2 is out of bounds for axis 0 with size 1` because `A[1:]` only has 1 row.
- **Does changing `D` change `A`?** Yes, because `B` is a slice (view) of `A` and `.T` on a 1D array returns another view sharing the exact same memory buffer.
- **One-line fix:** Change `B = A[1:, 2]` to `B = A[1:, 2].copy()`.

### Question 4
Consider an array containing the integers from 1 to 24.
- Write NumPy code to create this array.
- Determine which of the following shapes are valid for reshaping the array: `(2, 12)`, `(3, 8)`, `(4, 6)`, `(2, 3, 4)`, `(5, 5)`.
- Explain mathematically why some shapes are valid, and others are not.

````python
a = np.arange(1, 25)
shapes = [(2, 12), (3, 8), (4, 6), (2, 3, 4), (5, 5)]

for s in shapes:
    try:
        r = a.reshape(s)
        print(f"{s}: Valid -> shape {r.shape}")
    except ValueError:
        print(f"{s}: Invalid")
````

**Mathematical Reason:**  
Reshaping preserves the total number of elements $N = 24$. A target shape $(d_1, d_2, \dots, d_k)$ is valid iff $\prod_{i=1}^{k} d_i = 24$:
- Valid: $2 \times 12 = 24$, $3 \times 8 = 24$, $4 \times 6 = 24$, $2 \times 3 \times 4 = 24$.
- Invalid: $5 \times 5 = 25 \neq 24$.

### Question 6
Let:
$$A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}, \quad B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix}$$
- Calculate $A * B$ by hand.
- Calculate $A @ B$ by hand.
- Write NumPy statements for both operations.
- Why are the two results different?

````python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("A * B:\n", a * b)
print("A @ B:\n", a @ b)
````

### Question 8
A data analyst has the following array: `x = [10, 20, 30, 40]`.
- What happens to the ordinary mean if one entry is replaced by `np.nan`?
- Explain the purpose of `np.nanmean()` and why the data type of an array can matter during numerical computation.

````python
x = np.array([10, 20, 30, 40], dtype=float)
x[1] = np.nan

print("mean:", np.mean(x))
print("nanmean:", np.nanmean(x))
````

**Explanation:**
- **Ordinary mean:** Returns `nan` because IEEE 754 floating-point arithmetic propagates `NaN` through any sum/division.
- **`np.nanmean()`:** Computes the arithmetic mean while ignoring `NaN` entries (here `(10 + 30 + 40) / 3 = 26.67`).
- **Why `dtype` matters:** `np.nan` is strictly a float; integer arrays cannot store `np.nan` without throwing a `ValueError`, and smaller dtypes (like `int8` or `float16`) risk integer overflow or precision loss during large aggregations.

````python
a = np.array([10, 20, 30, 40])
b = a[1:3]
c = a[1:3].copy()
b[0] = 99
c[1] = 77

print("a:", a)
print("b:", b)
print("c:", c)
print("b shares memory with a:", np.shares_memory(a, b))
print("c shares memory with a:", np.shares_memory(a, c))
````

Buffer 2 (c):      [ 20 | 77 ]  (separate allocation)
```

Consider:

```
- Predict the output of both print statements.
- Explain how the Boolean mask selects elements.
- Why is `&` used here instead of the Python keyword `and`?
- Rewrite the condition to select values less than or equal to 10, or greater than or equal to 30.

````python
x = np.array([12, 25, 7, 31, 18, 40])
mask = (x > 10) & (x < 30)
print(mask)
print(x[mask])

m2 = (x <= 10) | (x >= 30)
print(x[m2])
````

**Explanation:**
- **Outputs:** `[ True  True False False  True False]` and `[12 25 18]`.
- **Selection:** NumPy maps the boolean array 1-to-1 against `x` and extracts only elements aligned with `True`.
- **Why `&` instead of `and`:** Python's `and` forces truth-value evaluation of the whole object (`bool(arr)`), raising `ValueError: The truth value of an array with more than one element is ambiguous`. `&` is overloaded in NumPy for elementwise logical AND.
- **Inverted condition:** `(x <= 10) | (x >= 30)` (or `~mask`).

### Question 15
A temperature sensor records the following readings in degrees Celsius:
`T = [28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7]`
- Calculate the mean and standard deviation using NumPy.
- Create a Boolean mask to identify readings greater than $40^\circ\text{C}$.
- Use `np.where()` to replace readings above $40^\circ\text{C}$ with the mean of the non-anomalous readings.
- Explain why a sensor anomaly should be investigated before automatically replacing the reading.

````python
t = np.array([28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7])

print(f"Raw mean: {np.mean(t):.2f}, std: {np.std(t):.2f}")

anom = t > 40
clean_mean = np.mean(t[~anom])
t_fixed = np.where(anom, clean_mean, t)

print("Mask:", anom)
print("Cleaned T:", np.round(t_fixed, 2))
````

**Why Investigate First:**  
A spike to $65.0^\circ\text{C}$ could be a real thermal runaway, fire, or hardware failure rather than bit-flip noise—auto-imputing it blindly can mask critical safety events and artificially deflate variance.

### Question 19
Create two arrays, each containing at least 100,000 numerical values. Compute the expression:
$$y = 3x^2 + 2x + 1$$
using (i) a Python loop and (ii) NumPy vectorized operations.
- Write both implementations.
- Check whether the two methods produce the same numerical results.
- Measure and record the execution time of each method using Python's `time` module.
- Repeat the experiment at three different array sizes and present your observations in a table.
- Explain why vectorization can be faster. Mention at least one limitation or caution when comparing execution times.

````python
from time import perf_counter

x1 = np.random.rand(100000)
x2 = np.random.rand(100000)

def exprL(x):
    y = [0.0] * len(x)
    for i in range(len(x)):
        y[i] = 3 * x[i]**2 + 2 * x[i] + 1
    return y

def exprN(x):
    return 3 * x**2 + 2 * x + 1

start = perf_counter()
y1 = exprL(x1)
y2 = exprL(x2)
loop_time = perf_counter() - start

start = perf_counter()
y1n = exprN(x1)
y2n = exprN(x2)
numpy_time = perf_counter() - start

print(f"Python loop (2x100k): {loop_time:.6f} s")
print(f"NumPy vectorized (2x100k): {numpy_time:.6f} s")
print("Results match:", np.allclose(y1, y1n) and np.allclose(y2, y2n))

print("\nSize     | Loop (s)   | NumPy (s)  | Speedup")
print("-" * 46)
for n in [1_000, 100_000, 1_000_000]:
    a = np.random.rand(n)
    t0 = perf_counter(); exprL(a); tl = perf_counter() - t0
    t0 = perf_counter(); exprN(a); tn = perf_counter() - t0
    print(f"{n:<8} | {tl:<10.6f} | {tn:<10.6f} | {tl/tn:.1f}x")
````

**Why Vectorization is Faster & Benchmarking Cautions:**
- **Speedup reason:** NumPy dispatches operations over contiguous typed memory blocks to pre-compiled C loops using CPU SIMD instructions (AVX/SSE), avoiding Python's per-iteration bytecode dispatch, dynamic type checks, and object boxing.
- **Benchmarking cautions:** For tiny arrays ($N < 100$), NumPy's C-API call and temporary intermediate buffer allocations (`3*x**2`, `2*x`, `+`) can actually make it slower than pure Python; also, single-run timings fluctuate due to CPU frequency scaling, OS background noise, and cold caches.
