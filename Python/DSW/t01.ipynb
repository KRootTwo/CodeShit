```python
f = ['apple', 'kiwi', 'fig', 'banana', 'watermelon']
sorted(f, key = lambda x: x[0])
```

```python
f = ['apple', 'kiwi', 'fig', 'banana', 'watermelon']
sorted(f, key = lambda x: x[-1])
```

```python
f = ['apple', 'kiwi', 'fig', 'banana', 'watermelon']
sorted(f, key = lambda x: x[1], reverse=True)
```
