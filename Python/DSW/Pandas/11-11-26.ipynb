```python
import pandas as pd
url = "https://datasets.imdbws.com/title.basics.tsv.gz"
df = pd.read_csv(url, sep="\t", low_memory=False)
ramayana = df[df['primaryTitle'].str.contains('Ramayana',case = False,na=False)]
#print(ramayana[['tconst','primaryTitle','startYear','runtimeMinutes','genres']])
rating_url="https://datasets.imdbws.com/title.ratings.tsv.gz"
rating_df = pd.read_csv(rating_url, sep="\t")
result = pd.merge(ramayana, rating_df, on='tconst',how="left")
print(result[['primaryTitle','startYear','averageRating','numVotes']])
```
