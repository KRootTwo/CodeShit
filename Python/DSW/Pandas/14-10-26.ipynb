## Reindexing

```python
fr = pd.DataFrame(np.arange(9).reshape((3, 3)),index = ['i', 'ii', 'iii'], columns= ['a', 'b', 'c'])
fr
```

```python
fr.reindex(['i', 'iii', 'ii','iv'], method='ffill')
```

```python
fr.reindex(['i', 'iii', 'ii','iv'], method='bfill')
```

```python
fr.reindex([1, 2, 3])
```

```python
fr = pd.DataFrame(np.arange(9).reshape((3, 3)),index = ['i', 'ii', 'iii'], columns= ['a', 'b', 'c'])
fr
```

```python
fr1 = fr.reindex(columns=['A', 'B'])
fr1
```

```python
fr2 = fr.reindex(['ii', 'A', '0', 'i'], axis=0)
fr2
```

```python
fr3 = fr.reindex(['C', 'A', 'D', 'B'], axis=1)
fr3
```

```python
fr4 = fr.reindex(['ii', 'A', '0', 'i'], axis=0, fill_value=7)
fr4
```

## Dropping entries from axis

```python
fr = pd.DataFrame(np.arange(12).reshape((3, 4)), columns = ['aee', 'bee', 'cee', 'dee'], 
                  index = ['A', 'B', 'C'])
fr
```

```python
fr.drop(columns=['aee', 'bee'])
```

```python
fr.drop(columns=['aee'], inplace=True)
fr
```

```python
fr.drop('A', axis=0)
```
