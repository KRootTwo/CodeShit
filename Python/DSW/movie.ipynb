```python
# Koustav Jana
import pandas as pd

df_m = pd.read_csv("title.basics.tsv.gz", sep = "\t", usecols = ["tconst", "primaryTitle", "startYear", "titleType"], low_memory =  False,
)

df_r = pd.read_csv("title.ratings.tsv.gz", sep="\t", usecols=["tconst", "averageRating", "numVotes"],)

movie = input("Search movie: ").strip()
if not movie:
    print("Empty title provided.")
else:
    m_title = df_m[df_m["primaryTitle"].str.contains(movie, case=False, na=False, regex=False)]
    found = m_title.merge(df_r, on="tconst", how="inner")

    if found.empty:
        print(f"No results for {movie}.")
    else:
        top10 = (
            found.sort_values(by = "numVotes", ascending = False)
            [["primaryTitle", "startYear", "averageRating", "numVotes"]].head(10)
        )
        print(f"\nResults for '{movie}':")
        print(top10.to_string(index=False))
```
