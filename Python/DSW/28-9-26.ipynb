map() function applies a given function to every element in an iterable.
    
The filter() function tests each element with a function that returns a boolean (True or False). It extracts only the elements that evaluate to True.
    
Unlike map() and filter(), reduce() is not globally built-in and must be imported from the functools module. It applies a function continuously to pairs of values to reduce the entire sequence to a single outcome.

```python
L = [1, 2, 3, 4]
result = list(map(lambda x: x + x, L))
result
```

```python
seq = [0, 1, 2, 3, 5, 8, 13]
res = list(filter(lambda x: x % 2 != 0, seq))
res
```

```python
import functools as f
L = [2, 1, 0, 2, 2, 0, 0, 2]
print(f.reduce(lambda x, y: x + y, L))
```

```python
ls = [1, 2, 3, 4, 6, 7, 9]

print(f.reduce(lambda x, y: x if x > y else y, ls))
```
