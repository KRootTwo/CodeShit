* Important foe exams
In Python, the four main types of function arguments are:
Positional Arguments
greet("Alice", 25)

Keyword Arguments
greet(age=25, name="Alice")

Default Arguments
greet("Alice")

Arbitrary (Variable-length) Arguments

```python
# b is default arg
def f(a, b=5):
    return a + b
f(a=5)
```

```python
def f(a, b)
    print(a + b)
```

```python
f(a = 3, b = 4)
```

```python
f(a = 3, 4)
```

```python
f(4, b = 5)
```

```python
f(a = 7, b = 9)
```

```python
def f(a = 2, b = 3):
    return a + b
f(a = 4, b = 9)
```

```python
def summ(*a):
    s = 0
    for i in a:
        s = s + i
    return s
summ(1, 2, 3, 4, 5, 6)
```

```python
def maksimum(*a):
    m = a[0]
    for i in a:
        if i > m:
            m = i
    return m
maksimum(10, 1, 2, 7, 100)
```

```python
def p(**a):
    for i, j in a.items():
        print(i, j)
p(rahul = 8, col = 90)
```

```python
def p(**a):
    for i in a.values():
        print(i)
p(rahul = 8, cole = 90)
```
