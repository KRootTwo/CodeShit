```python
D = {'A' : 1, 'B' : 2, 'C' : 1, 'D' : 2}
```

```python
D1 = dict()
for i, j in D.items():
    if j not in D1.keys():
        D1[j] = list()
    D1[j].append(i)
D1
```

```python
D1 = {1: ['A', 'C'], 2: ['B', 'D'], 3 : 'G'}
D2 = dict()
for i, j in D1.items():
    if j.__iter__():
        for k in j:
            D2[k] = i
    else :
        D2[j] = i
D2
```

```python
R = [4, 5, 6, 10, 11, 12, 13, 3]
k = []
for i in range(len(R)):
    l = [R[i]]
    for j in range(i, len(R) - 1):
        if R[j + 1] - R[j] == 1:
            l.append(R[j + 1])
        else :
            break
    k.append(l)
max = []
print(k)
for t in k:
    if (len(t) > len(max)) :
        max = t
max
```
