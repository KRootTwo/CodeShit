**Write a python script to solve the following problem using loops and functions:**

Write a function `classify_character(ch)` that takes a single character and prints whether it is:
- a vowel (`a`, `e`, `i`, `o`, `u`, case-insensitive)
- a consonant
- a digit
- a whitespace
- a special symbol

Use nested `if-elif-else`. Test with `'A'`, `'z'`, `'7'`, `' '`, `'@'`.

````python
def classify_character(c):
    a = ord(c)
    if a >= 48 and a <= 57:
        print(f"{c} is a digit")
    elif a == 32:
        print(f"{c} is a whitespace")
    elif ((a >= 65 and a <= 90) or (a >= 97 and a <= 122)) :
        vow = 'aeiouAEIOU'
        if c in vow:
            print(f"{c} is a vowel")
        else:
            print(f"{c} is a consonant")
    else :
        print(f"{c} is a special character")
classify_character('A')
classify_character('z')
classify_character('7')
classify_character(' ')
classify_character('@')
````

### Question 1(b)
Write a function `electricity_bill(units)` that calculates the bill using slab rates:
- **First 100 units:** Rs 3/unit
- **Next 100 units:** Rs 5/unit
- **Next 300 units:** Rs 7/unit
- **Above 500 units:** Rs 10/unit

Use `if-elif-else`. Return the total bill.

````python
def electricity_bill(units) :
    if units < 0 :
        return 0
    elif units <= 100 :
        return units * 3
    elif units > 100 and units <= 200 :
        return 100 * 3 + (units - 100) * 5
    elif units > 200 and units <= 500 :
        return 100 * 3 + 100 * 5 + (units  - 200) * 7
    else :
        return 100 * 3 + 100 * 5 + 300 * 7 + (units - 500) * 10
electricity_bill(900)
````

### Question 1(c)
Write function `calc_armstrong_numbers()` to compute all Armstrong numbers for $x$, $y$, and $z$ as digits of the number. This deals with three-digit Armstrong numbers.

*[Example: $153 = 1 \times 100 + 5 \times 10 + 3 = 1^3 + 5^3 + 3^3$]*

````python
def calc_armstrong_numbers() :
    arm = []
    l = 3
    for i in range(100, 1000):
        x = (i // 100)
        y = (i // 10) % 10
        z = i % 10

        if (x ** l + y ** l + z ** 3) == i :
            arm.append(i)
    return arm
calc_armstrong_numbers()
````

### Question 1(e)
Write a program that uses a `for-else` loop to check if a given list contains any duplicate elements.

If a duplicate is found, `break` and print `"Duplicate found"`. If no duplicate, the `else` block prints `"All elements are unique"`.

````python
def check_dupes(items):
    s = set()
    for i in items:
        if i in s:
            break
        s.add(i)
    else:
        print("All elements are unique")
````

### Question 3
Write a function `collatz_steps(n)` that returns the number of steps to reach 1 using the rules:
- If $n$ is even: $n = n // 2$
- If $n$ is odd: $n = 3 \times n + 1$

Use a `while` loop. Test with $n = 6$ *(expected: 8 steps)*.

````python
def collatz_steps(n):
    s = 0
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        s += 1
    return s

collatz_steps(6)
````

### Question 4
Write a function `fizzbuzz_plus(n)` that prints numbers from 1 to $n$ with these rules:
- Multiples of 3 $\rightarrow$ `"Fizz"`
- Multiples of 5 $\rightarrow$ `"Buzz"`
- Multiples of 7 $\rightarrow$ `"Bang"`
- Multiples of 3 and 5 $\rightarrow$ `"FizzBuzz"`
- Multiples of 3 and 7 $\rightarrow$ `"FizzBang"`
- Multiples of 5 and 7 $\rightarrow$ `"BuzzBang"`
- Multiples of 3, 5, and 7 $\rightarrow$ `"FizzBuzzBang"`
- Otherwise $\rightarrow$ the number itself

````python
def fizzbuzz_plus(n):
    for i in range(1, n + 1):
        if i % 3 == 0: print("Fizz")
        elif i % 5 == 0: print("Buzz")
        elif i % 7 == 0: print("Bang")
        else : print(i)

fizzbuzz_plus(15)
````
