#### Student Marks Analysis
Store the marks of 8 students in a list. Write a program to display the highest marks, lowest marks, average marks, and the number of students who scored above the average.

```python
m = [85, 92, 78, 88, 95, 70, 82, 90]
hi = max(m)
lo = min(m)
avg = sum(m) / len(m)
cnt = sum(1 for x in m if x > avg)
print("Highest:", hi)
print("Lowest:", lo)
print("Average:", avg)
print("Above average:", cnt)
```

#### Remove Duplicate Entries
A list contains repeated student registration numbers. Write a program to create a new list containing only unique registration numbers while preserving their original order.

**Example:**
- **Input:** `[101, 103, 101, 105, 103, 108]`
- **Output:** `[101, 103, 105, 108]`

```python
l = [101, 103, 101, 105, 103, 108]
res = []
for i in l:
    if i not in res:
        res.append(i)
print(res)
```

#### Word Frequency Counter
Accept a sentence from the user and create a dictionary that stores each word and its frequency. Display the words that occur more than once.

**Example:**
- **Input:** `data science uses data and science`
- **Output:** `{'data': 2, 'science': 2, 'uses': 1, 'and': 1}`

```python
s = input("Enter sentence: ")
d = {}
for w in s.lower().split():
    d[w] = d.get(w, 0) + 1
print(d)
print("Repeated words:", [w for w in d if d[w] > 1])
```

#### Shopping Bill Generator
Create a dictionary containing the names of five products and their prices. Accept the names and quantities of three products purchased by a customer. Calculate and display the total bill.

```python
p = {"Pen": 10, "Book": 45, "Backpack": 300, "Pencil": 5, "Eraser": 3}
t = 0
for _ in range(3):
    n = input("Product: ")
    q = int(input("Qty: "))
    if n in p:
        t += p[n] * q
    else:
        print("Not found")
print("Total bill: Rs", t)
```

#### Tuple-Based Coordinate Distance
Store two points as tuples, such as $(x_1, y_1)$ and $(x_2, y_2)$. Write a program to calculate the Euclidean distance between the points using the formula:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

```python
import math
p1 = (2, 3)
p2 = (7, 8)
d = math.sqrt((p2[0]-p1[0])**2 + (p2[1]-p1[1])**2)
print("Dist:", d)
```

#### List Rotation by One Position
Write a program to move the last element of a list to the first position. The remaining elements should shift one position to the right.

**Example:**
- **Input:** `[10, 20, 30, 40, 50]`
- **Output:** `[50, 10, 20, 30, 40]`

```python
l = [10, 20, 30, 40, 50]
l = [l[-1]] + l[:-1]
print(l)
```

#### Inventory Management
Create a dictionary containing the names of five items and their available quantities. Accept the name and quantity of an item sold. Update the inventory and display the modified dictionary. If the requested quantity is unavailable, display an appropriate message.

```python
inv = {"Pen": 50, "Book": 30, "Backpack": 10, "Pencil": 100, "Eraser": 75}
n = input("Item: ")
q = int(input("Qty: "))
if n in inv and inv[n] >= q:
    inv[n] -= q
    print(inv)
else:
    print("Unavailable")
```

#### What is the difference between a list and a tuple? Give one practical example of each.

Ans:- A list is mutable (can be changed after creation) and is written with square brackets []. A tuple is immutable (cannot be changed) and is written with parentheses ()

```python
# List
marks = [70, 80, 90]
marks[0] = 75          # allowed
print(marks)           # [75, 80, 90]

# Tuple
pt = (3, 5)
# pt[0] = 4            # ERROR: tuple is immutable
print(pt)              # (3, 5)
```

#### Explain how a set removes duplicate elements. Demonstrate this using a list of your choice.

Ans:- A set stores only unique values. When an iterable is passsed (like a list) to set(), Python inserts each element if and only if it is not already present, so duplicates are discarded.

```python
l = [1, 2, 2, 3, 4, 4, 4, 5]
s = set(l)
s
```

#### What is the difference between `append()` and `extend()` in Python lists? Demonstrate both using the same list.

Ans:- append(x) adds x as a single element at the end.

extend(iterable) adds each element of the iterable individually at the end.

```python
l = [1, 2, 3]
l.append([4, 5])
l
```

```python
l = [1, 2, 3]
l.extend([4, 5])
l
```
