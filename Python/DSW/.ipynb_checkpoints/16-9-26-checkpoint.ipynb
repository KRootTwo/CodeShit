```python
# 1
name = '''I am Amit pani. 
        My village is Pahi Sahi, 
        My friend is Aspati'''
pat = r'\b[Pp]\w{0,3}i\b'
re.findall(pat, name)
```

```python
# 2
f = '100klh564abc365bg'

l = re.split('[a-z]+', f)
if (l[0].isnumeric()): 
    d = int(l[0])

for i in range(1, len(l)):
    if (l[i].isnumeric()):
        d = max(int(l[i]), d)
        
print(d)
    
```

```python
# 3
num = '''
        234-678-9876
        2345-876-987
        6  5-872-0987
        234.987.8902
        2e3-345-8765'''

pat = r'\b\d{3}-\d{3}-\d{4}\b'
re.findall(pat, num)
```

```python
# 4
pat = r'[a-z{1,}A-Z{1,}0-9{1,}@%+-]{1,}'
pat = r'[a-zA-Z0-9@%+-]+'
pas = '''Ramrehman+7
        Ramrehman+'''

pa = re.findall(pat, pas)
for p in pa:
    if '@' in p or '%' in p or '+' in p or '-' in p:
        print("strong")

```

```python
p = '''Ramrehman+7'''
r = '''Ramrehman+'''
# p = re.split('\w+')

def isVal(p):
    
    p1 = r'[0-9]{1,}'
    p2 = r'[a-z]{1,}'
    p3 = r'[A-Z]{1,}'
    p4 = r'[*@+-]{1,}'

    if re.search(p1, p) and re.search(p2, p) and re.search(p3, p) and re.search(p4, p):
        return 'strong'
    return 'weak'

print(isVal(p))
print(isVal(r))
```

## String Formatting

```python
x = 12
print('%d' % x)
```

```python
x = 679.123123123
print("int is - %d float is - %2.3f string is - %s" % (x, x, x))
```

```python
print("%8s" % 'Hey')
print("%8s" % 'Hello World')
```

```python
# Using curly brackets
print("{} and {}".format("Ashis", "Namish"))
# Positional Argument
print("{1} and {0}".format("Ashis", "Namish"))
# Keyword 
print("{a} and {b}".format(a = "Ashis", b = "Namish"))
```
