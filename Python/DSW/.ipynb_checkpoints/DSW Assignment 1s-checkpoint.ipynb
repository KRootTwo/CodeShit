```python
# Q1
x = 10
y = 3

print(x / y)
print(x // y)
print(x % y)
print(x ** y)
print(x > y and y > 5)
print(x > y or y > 5)
print(not x > y)

# Modified values so all three logical expressions change result
x = 3
y = 10
print(x > y and y > 5)
print(x > y or y > 5)
print(not x > y)
```

```python
# Q3
text = " Python Programming is Powerful "

print("Original length:", len(text))

cleaned = text.strip()
print("Cleaned length:", len(cleaned))

print("First occurrence of 'Python':", cleaned.find("Python"))
print("Last occurrence of 'o':", cleaned.rfind("o"))
print("Number of occurrences of 'o':", cleaned.count("o"))
print("Number of words:", len(cleaned.split()))
print("Starts with 'Python':", cleaned.startswith("Python"))
```

```python
# Q4
try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    result = a / b
    print("Result:", result)
except ZeroDivisionError:
    print("Denominator cannot be zero.")
finally:
    print("Program execution completed.")
```

```python
# Q5
try:
    number = int(input("Enter an integer: "))
except ValueError:
    print("Invalid input. Please enter an integer.")
else:
    print("Square:", number ** 2)
```

```python
# Q6
numbers = [10, 20, 30, 40, 50]

try:
    index = int(input("Enter an index: "))
    divisor = int(input("Enter a divisor: "))
    result = numbers[index] / divisor
except ValueError:
    print("Invalid input. Please enter integers only.")
except IndexError:
    print("Index is outside the list.")
except ZeroDivisionError:
    print("Divisor cannot be zero.")
else:
    print("Result:", result)
    print("Operation completed successfully.")
finally:
    print("Program has finished.")
```

```python
# Q7
filename = input("Enter the name of the text file: ")

try:
    file = open(filename, "r")
    contents = file.read()
except FileNotFoundError:
    print("File not found.")
else:
    print(contents)
finally:
    print("File operation completed.")
```

```python
# Q9
pairs = [
    ((8, 1), (8, 1), (8, 1)),
    ((16, 2), (16, 2), (16, 2)),
    ((32, 3), (32, 3), (32, 3))
]

for (a1, b1), (a1_2, b2), (a2, b1_2) in pairs:
    print(a1 >> b1, a1_2 >> b2, a2 >> b1_2)
```

```python
# Q10
students = ["Rahul", "Aman", "Priya", "Sneha", "Rohan"]

with open("students.txt", "w") as file:
    for student in students:
        file.write(student + "\n")

with open("students.txt", "r") as file:
    print(file.read())

# Running the program again in "w" mode overwrites the previous contents.
```

```python
# Q11
initial_marks = [
    "Rahul - 85\n",
    "Aman - 78\n",
    "Priya - 92\n",
    "Sneha - 88\n",
    "Rohan - 81\n"
]

with open("marks.txt", "w") as file:
    file.writelines(initial_marks)

additional_marks = [
    "Ankit - 76\n",
    "Neha - 90\n",
    "Karan - 84\n"
]

with open("marks.txt", "a") as file:
    file.writelines(additional_marks)

with open("marks.txt", "r") as file:
    print(file.read())
```

```python
# Q12
story = "Python is easy to learn.\nPython is powerful.\nPython is widely used in AI.\n"

with open("story.txt", "w") as file:
    file.write(story)

with open("story.txt", "r") as file:
    print("Initial position:", file.tell())

    print("First 10 characters:", file.read(10))
    print("Position after reading 10 characters:", file.tell())

    file.seek(0)
    print("First line:", file.readline().strip())

    file.seek(10)
    print("Characters from position 10:", file.read(20))
```

```python
# Q13
records = [
    "Rahul,101,85\n",
    "Aman,102,78\n",
    "Priya,103,92\n",
    "Sneha,104,88\n",
    "Rohan,105,81\n"
]

with open("student.txt", "w") as file:
    file.writelines(records)

with open("student.txt", "r") as file:
    all_records = file.readlines()
    for record in all_records:
        print(record.strip())

    file.seek(0)
    print("First record:", file.readline().strip())
    print("Current position:", file.tell())

with open("student.txt", "a") as file:
    file.write("Ankit,106,76\n")
    file.write("Neha,107,90\n")

with open("student.txt", "r") as file:
    print("\nFinal contents:")
    print(file.read())
```

```python
# Q14
import re

data = [
    "student@gmail.com",
    "abc@soa.ac.in",
    "9876543210",
    "student123",
    "test@example.com"
]

pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

emails = []

for item in data:
    if re.search(pattern, item):
        emails.append(item)

print("Valid email addresses:")
print(emails)
```

```python
# Q15
def student_report(name, roll_number, branch="CSE", **kwargs):
    print("STUDENT REPORT")
    print("-" * 30)
    print(f"Name      : {name}")
    print(f"Roll No.  : {roll_number}")
    print(f"Branch    : {branch}")

    for key, value in kwargs.items():
        print(f"{key.capitalize():<10}: {value}")

student_report(
    "Rahul",
    101,
    branch="CSE",
    cgpa=8.7,
    city="Bhubaneswar"
)
```

```python
# Q16
def data_summary(*args, **kwargs):
    if not args:
        print("No numerical values supplied.")
        return

    print("Number of values:", len(args))
    print("Minimum value:", min(args))
    print("Maximum value:", max(args))
    print("Total:", sum(args))
    print("Average:", sum(args) / len(args))

    if "dataset" in kwargs:
        print("Dataset:", kwargs["dataset"])
    if "unit" in kwargs:
        print("Unit:", kwargs["unit"])

data_summary(
    10, 20, 30, 40, 50,
    dataset="Marks",
    unit="percentage"
)
```

```python
# Q17
a = 25
b = 40
c = 17

square = lambda x: x ** 2
larger = lambda x, y: x if x > y else y
is_even = lambda x: x % 2 == 0

print("Square of a:", square(a))
print("Larger of a and b:", larger(a, b))
print("Is c even?:", is_even(c))
```

```python
# Q18
# calculator.py
calculator_code = '''
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

if __name__ == "__main__":
    print("Addition:", add(10, 5))
    print("Subtraction:", subtract(10, 5))
    print("Multiplication:", multiply(10, 5))
    print("Division:", divide(10, 5))
'''

with open("calculator.py", "w") as file:
    file.write(calculator_code)

# main.py
main_code = '''
from calculator import add, subtract, multiply, divide

print("Addition:", add(20, 10))
print("Subtraction:", subtract(20, 10))
print("Multiplication:", multiply(20, 10))
print("Division:", divide(20, 10))
'''

with open("main.py", "w") as file:
    file.write(main_code)

print("calculator.py and main.py created.")
```
