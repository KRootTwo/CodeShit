## all & any

* any() - tests whether on or more values in a array in True
* all() - checks if every value is True

```python
any(np.array([False, False, True, False, False]))
```

```python
all(np.array([False, False, True, False, False]))
```

```python
mk = np.array([10, 20, 60])
```

```python
all(mk > 30)
```

```python
any(mk > 30)
```

## NumPy sorting

* list.sort() sorts a list in place and returns None.
* sorted() returns a new sorted list, leaving the original unchanged.

    Similarly
* a.sort() sorts the array in place and returns None.
* np.sort(a) returns a sorted copy of the array.

## Unique Values

```python
arr1 = np.array([2, 1, 3, 5, 4])
arr2 = np.array([9, 6, 7, 3, 8])
```

```python
np.isin(arr1, arr2)
```

## File Handling in NumPy

```python
import numpy as np
r = np.random.default_rng()
a = r.standard_normal(10)
np.save('abc', a)
```

```python
np.load('abc.npy')
```

```python
np.savez('def.npz', x = a, y = a)
unzip = np.load('def.npz')
unzip['x']
```

```python
l = np.array([-1, 1])
s = np.random.choice(l, size=1000)
w = np.cumsum(s)
w
```

```python
import matplotlib.pyplot as plt
plt.style.use("dark_background")
plt.plot(w, color="yellow")
```
