**Q1.** Write down the variable and its values after execution of each line of the program. Predict all four outputs. Whether changing `D` will change `A`? Modify only one line of the program so that changing `D` does NOT change `A`.

````python
A = np.arange(1, 7).reshape(2, 3)
B = A[1:, 2].copy()
D = B.T
D[0] = 100

print(A)
print(B)
print(D)
````

- `A` becomes `[[1, 2, 3], [4, 5, 6]]` (shape 2x3).
- `B` becomes `[6]` (1D view of element at row 1, col 2).
- `C = A[1:][2]` throws `IndexError` because `A[1:]` only has 1 row (index 0).
- Yes, changing `D` changes `A` because `B` is a view of `A` and `B.T` also shares the same memory.
- Fix: change line 2 to `B = A[1:, 2].copy()`.

**Q3.** Given a matrix representing marks of a student in different subjects (columns) and different semesters (rows), identify where the mark is 50+ in at least two subjects. Write the algorithm, draw a flowchart, take your semester 1, semester 2, and expected semester 3 marks, write the code, and execute this program.

````python
m = np.array([
    [76, 81, 44, 85, 72],
    [47, 83, 41, 78, 80],
    [84, 87, 79, 89, 82]
])

c = np.sum(m >= 50, axis=1)
r = np.where(c >= 2)[0] + 1
print("Subjects 50+ per sem:", c)
print("Qualified semesters:", r)
````

**Algorithm:**
1. Start and store semester marks in a 2D array `m` (rows = semesters, cols = subjects).
2. Check `m >= 50` to get a boolean matrix.
3. Sum along `axis=1` to count subjects with 50+ marks in each semester.
4. Find semester indices where count >= 2.
5. Print the semester numbers and stop.

**Flowchart:**
```text
(Start) -> [Input matrix m] -> [b = m >= 50] -> [c = sum(b, axis=1)] -> [Find rows with c >= 2] -> [Print semesters] -> (End)
```

**Q4.** Consider an array containing the integers from 1 to 24. Write NumPy code to create this array. Determine which of the following shapes are valid for reshaping the array: `(2,12)`, `(3,8)`, `(4,6)`, `(2,3,4)`, `(5,5)`. Explain mathematically why some shapes are valid, and others are not.

````python
a = np.arange(1, 25)
print(a.reshape(2, 12).shape)
print(a.reshape(3, 8).shape)
print(a.reshape(4, 6).shape)
print(a.reshape(2, 3, 4).shape)
````

The total number of elements is 24. Reshaping only works when the product of the new dimensions equals 24:
- Valid: `2*12 = 24`, `3*8 = 24`, `4*6 = 24`, `2*3*4 = 24`.
- Invalid: `5*5 = 25 != 24`, so `(5, 5)` throws a `ValueError`.

**Q5.** Suppose `A.shape` is `(4, 7, 1)`. For each of the following shapes, state whether NumPy broadcasting is possible when combining `A` elementwise with an array of that shape: `(7,1)`, `(4,7)`, `(1,7,1)`, `(4, 1,5)`, `(3, 7, 1)`. If broadcasting is possible, give the shape of the output.

````python
a = np.ones((4, 7, 1))
for s in [(7, 1), (4, 7), (1, 7, 1), (4, 1, 5), (3, 7, 1)]:
    try:
        b = a + np.ones(s)
        print(s, "-> Possible, output shape:", b.shape)
    except ValueError:
        print(s, "-> Not possible")
````

Comparing dimensions from right to left against `(4, 7, 1)`:
- `(7, 1)` -> Possible, output `(4, 7, 1)`
- `(4, 7)` -> Not possible (trailing `7` and `1` clash)
- `(1, 7, 1)` -> Possible, output `(4, 7, 1)`
- `(4, 1, 5)` -> Possible, output `(4, 7, 5)`
- `(3, 7, 1)` -> Not possible (`3` and `4` don't match)

**Q6.** Let $A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$ and $B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix}$. Calculate $A * B$ and $A @ B$ by hand. Write NumPy statements for both operations. Why are the two results different?

````python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(a * b)
print(a @ b)
````

- $A * B = \begin{bmatrix} 5 & 12 \\ 21 & 32 \end{bmatrix}$
- $A @ B = \begin{bmatrix} 1(5)+2(7) & 1(6)+2(8) \\ 3(5)+4(7) & 3(6)+4(8) \end{bmatrix} = \begin{bmatrix} 19 & 22 \\ 43 & 50 \end{bmatrix}$
- `*` multiplies corresponding elements one by one, whereas `@` performs standard matrix multiplication (row-by-column dot products).

**Q7.** Consider $A = \begin{bmatrix} 2 & 4 & 6 \\ 1 & 3 & 5 \\ 7 & 8 & 9 \end{bmatrix}$. Without initially executing code, calculate `np.sum(A, axis=0)`, `np.sum(A, axis=1)`, `np.mean(A, axis=0)`, and `np.mean(A, axis=1)`. Verify using NumPy and explain what each axis operation does.

````python
a = np.array([[2, 4, 6], [1, 3, 5], [7, 8, 9]])

print(np.sum(a, axis=0))
print(np.sum(a, axis=1))
print(np.mean(a, axis=0))
print(np.mean(a, axis=1))
````

- `np.mean(A, axis=1)` = `[4.0, 3.0, 8.0]` (average of each row)

**Q8.** A data analyst has the array `x = [10, 20, 30, 40]`. What happens to the ordinary mean if one entry is replaced by `np.nan`? Explain the purpose of `np.nanmean()` and why the data type of an array can matter during numerical computation.

````python
x = np.array([10.0, 20.0, 30.0, 40.0])
x[2] = np.nan

print(np.mean(x))
print(np.nanmean(x))
````

- Replacing an entry with `np.nan` makes `np.mean(x)` return `nan` because any arithmetic with `NaN` evaluates to `NaN`.
- `np.nanmean()` skips `NaN` values and calculates the average of the remaining valid numbers (`(10+20+40)/3 = 23.33`).
- Data type matters because `np.nan` only works in floating-point arrays (assigning it to an integer array throws an error), and small integer/float types can overflow or lose decimal precision.

**Q9.** Predict the output and explain each step for `a = np.array([2.8, 3.9, -1.7])`, `b = a.astype(int)`. Would converting these values to integers round them to the nearest integer? Also discuss what happens when numeric values and strings are combined in a single NumPy array.

````python
a = np.array([2.8, 3.9, -1.7])
b = a.astype(int)
print(b)
print(b.dtype)

c = np.array([1, 2.5, "abc"])
print(c, c.dtype)
````

- Output: `[ 2  3 -1]` and `int64`.
- No, `astype(int)` does not round—it truncates the decimal part toward zero (`2.8` becomes `2`, `-1.7` becomes `-1`).
- When numbers and strings are combined in one array, NumPy converts all elements into strings (`<U32`) since all elements in an ndarray must share the same data type.

**Q10.** Predict all three outputs before running the code (`a = np.array([10, 20, 30, 40])`, `b = a[1:3]`, `c = a[1:3].copy()`, `b[0] = 99`, `c[1] = 77`). Explain why modifying `b` affects `a` while modifying `c` does not, and draw a simple memory diagram.

````python
a = np.array([10, 20, 30, 40])
b = a[1:3]
c = a[1:3].copy()
b[0] = 99
c[1] = 77

print(a)
print(b)
print(c)
````

- Outputs: `a = [10 99 30 40]`, `b = [99 30]`, `c = [20 77]`.
- Slicing `a[1:3]` creates a view pointing to `a`'s memory, so editing `b` edits `a`. `.copy()` creates a separate copy in memory, so `c` is independent.
```text
a ---> [ 10 | 99 | 30 | 40 ]
              ^    ^
b ------------|----|

c ---> [ 20 | 77 ]
```

**Q11.** Consider `x = np.array([12, 25, 7, 31, 18, 40])` and `mask = (x > 10) & (x < 30)`. Predict the output of `print(mask)` and `print(x[mask])`. Explain how the Boolean mask selects elements, why `&` is used instead of `and`, and rewrite the condition for values `<= 10` or `>= 30`.

````python
x = np.array([12, 25, 7, 31, 18, 40])
mask = (x > 10) & (x < 30)
print(mask)
print(x[mask])

m2 = (x <= 10) | (x >= 30)
print(x[m2])
````

- Outputs: `[ True  True False False  True False]` and `[12 25 18]`.
- `x[mask]` returns only the elements of `x` at positions where `mask` is `True`.
- `&` performs elementwise logical AND across arrays, whereas Python's `and` expects a single boolean and raises a `ValueError` on arrays.
- Rewritten condition: `(x <= 10) | (x >= 30)`.

**Q14.** Given $M = \begin{bmatrix} 78 & 85 & 90 \\ 45 & 55 & 49 \\ 88 & 92 & 95 \\ 32 & 40 & 38 \end{bmatrix}$ (rows = students, cols = subjects): create `M`, find student averages and subject averages, identify students with average < 50, and identify students who pass (at least 40 in every subject). Explain why array operations are more convenient.

````python
M = np.array([
    [78, 85, 90],
    [45, 55, 49],
    [88, 92, 95],
    [32, 40, 38]
])

sa = np.mean(M, axis=1)
sub = np.mean(M, axis=0)
low = np.where(sa < 50)[0]
p = np.where(np.all(M >= 40, axis=1))[0]

print("Student avg:", sa)
print("Subject avg:", sub)
print("Avg < 50 student indices:", low)
print("Passed student indices:", p)
````

Array operations (`axis=0`, `axis=1`, `np.all`) let us compute averages and check pass/fail conditions across the whole table in single lines without writing nested loops.

**Q15.** For `T = [28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7]`, find mean and std dev, create a Boolean mask for readings > $40^\circ\text{C}$, use `np.where()` to replace them with the mean of non-anomalous readings, and explain why an anomaly should be investigated before replacing.

````python
T = np.array([28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7])
print("Mean:", np.mean(T))
print("Std:", np.std(T))

m = T > 40
avg = np.mean(T[~m])
T2 = np.where(m, avg, T)
print("Mask:", m)
print("Replaced T:", T2)
````

A sudden spike like `65.0` could indicate an actual overheating event or hardware fault rather than a random glitch; replacing it automatically can hide real safety warnings.

**Q17.** For `P = [450, 1200, 800, 2500, 1500, 600]`, a 10% discount applies if purchase >= 1000. Create a Boolean mask for eligible customers, use `np.where()` to calculate final amounts, find total revenue before and after discounts, and explain how this scales to thousands of customers.

````python
P = np.array([450, 1200, 800, 2500, 1500, 600])
m = P >= 1000
f = np.where(m, P * 0.9, P)

print("Mask:", m)
print("Final:", f)
print("Before:", np.sum(P))
print("After:", np.sum(f))
````

Because `P >= 1000` and `np.where()` are vectorized, the exact same 2 lines work for 10,000+ customers without changing any code or writing loops.

**Q18.** Construct two NumPy arrays (at least 5 numbers each) with the same mean but different variances, verify with `np.mean()` and `np.var()`, explain what variance tells us, and construct two arrays with the same mean and same variance but different elements.

````python
a = np.array([8, 9, 10, 11, 12])
b = np.array([0, 5, 10, 15, 20])
print(np.mean(a), np.var(a))
print(np.mean(b), np.var(b))

c = np.array([1, 2, 3, 4, 5])
d = np.array([3 - np.sqrt(5), 3, 3, 3, 3 + np.sqrt(5)])
print(np.mean(c), np.var(c))
print(np.mean(d), np.var(d))
````

- Hand means: `a = (8+9+10+11+12)/5 = 10`, `b = (0+5+10+15+20)/5 = 10`.
- Variance shows how spread out the numbers are around the mean (`a` is tightly clustered with variance `2`, while `b` is widely spread with variance `50`).
- Arrays `c` and `d` both have mean `3.0` and variance `2.0` despite having different element values.

**Q19.** Create two arrays, each containing at least 100,000 numerical values. Compute $y = 3x^2 + 2x + 1$ using (i) a Python loop and (ii) NumPy vectorized operations. Check if both produce the same results, measure execution time using `time`, repeat at three different array sizes in a table, and explain why vectorization is faster along with one limitation/caution.

````python
from time import perf_counter
x1 = np.random.rand(100000)
x2 = np.random.rand(100000)

def exprL(x):
    y = [0.0] * len(x)
    for i in range(len(x)):
        y[i] = 3 * x[i]**2 + 2 * x[i] + 1
    return y
def exprN(x):
    return 3 * x**2 + 2 * x + 1

start = perf_counter()
y1 = exprL(x1)
y2 = exprL(x2)
loop_time = perf_counter() - start

start = perf_counter()
y1n = exprN(x1)
y2n = exprN(x2)
numpy_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")

print(y1 == y1n)
print(y2 == y2n)

a1 = np.random.rand(100)
a2 = np.random.rand(1000)
a3 = np.random.rand(10000)

start = perf_counter()
b1 = exprL(a1)
b2 = exprL(a2)
b3 = exprL(a3)
loop_time = perf_counter() - start

start = perf_counter()
c1 = exprN(a1)
c2 = exprN(a2)
c3 = exprN(a3)
loop_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")
````

| Array Size | Loop Time (s) | NumPy Time (s) |
| :--- | :--- | :--- |
| 100 | ~0.00007 | ~0.00002 |
| 1,000 | ~0.00065 | ~0.00003 |
| 10,000 | ~0.00680 | ~0.00009 |

- Vectorization runs compiled C loops over contiguous memory without Python interpreter overhead on every iteration.
- Caution: For very small arrays, NumPy function call overhead and temporary array allocations can make it slower than a basic Python loop, and single-run timings vary with CPU background load.
