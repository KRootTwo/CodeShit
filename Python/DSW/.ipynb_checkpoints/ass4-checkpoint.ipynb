**Q1.** Trace variable values after each line, predict all 4 outputs, check if changing D changes A, and modify one line so changing D does not change A.

```python
A = np.arange(1, 7).reshape(2, 3)
B = A[1:, 2].copy()
D = B.T
D[0] = 100

print(A)
print(B)
print(D)
```

A is [[1, 2, 3], [4, 5, 6]] and B is [6] (view of A[1, 2]). Line 3 (C = A[1:][2]) gives IndexError since A[1:] has only 1 row. Yes, changing D changes A because B and D are views sharing memory with A. Changing line 2 to B = A[1:, 2].copy() prevents this.

**Q3.** Given a semester-by-subject marks matrix, identify semesters with 50+ in at least two subjects. Write the algorithm, flowchart, and execute with Sem 1, Sem 2, and expected Sem 3 marks.

```python
m = np.array([
    [76, 81, 44, 85, 72],
    [47, 83, 41, 78, 80],
    [84, 87, 79, 89, 82]
])

c = np.sum(m >= 50, axis=1)
r = np.where(c >= 2)[0] + 1
print(c)
print(r)
```

Algorithm:
1. Store semester marks in matrix m.
2. Compare m >= 50 and sum across rows (axis=1) to count passed subjects per semester.
3. Find rows where count >= 2 and print semester numbers.

Flowchart: Start -> Input m -> c = sum(m >= 50, axis=1) -> Find c >= 2 -> Print semesters -> End

**Q4.** Create an array from 1 to 24. Check which shapes among (2,12), (3,8), (4,6), (2,3,4), (5,5) are valid and explain why.

```python
a = np.arange(1, 25)
print(a.reshape(2, 12).shape)
print(a.reshape(3, 8).shape)
print(a.reshape(4, 6).shape)
print(a.reshape(2, 3, 4).shape)
```

Total elements = 24, so the product of dimensions must equal 24. (2,12), (3,8), (4,6), and (2,3,4) all multiply to 24 and are valid, whereas (5,5) gives 25 and is invalid.

**Q5.** If A.shape is (4, 7, 1), check broadcasting compatibility and output shapes with (7,1), (4,7), (1,7,1), (4,1,5), and (3,7,1).

```python
a = np.ones((4, 7, 1))
for s in [(7, 1), (4, 7), (1, 7, 1), (4, 1, 5), (3, 7, 1)]:
    try:
        b = a + np.ones(s)
        print(s, b.shape)
    except ValueError:
        print(s, "Not possible")
```

Matching trailing dimensions with (4, 7, 1): (7,1) -> (4,7,1), (4,7) -> Not possible (7 != 1), (1,7,1) -> (4,7,1), (4,1,5) -> (4,7,5), (3,7,1) -> Not possible (3 != 4).

**Q6.** For $A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$ and $B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix}$, compute A*B and A@B by hand and in NumPy, and explain why they differ.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(a * b)
print(a @ b)
```

By hand: A*B = [[5, 12], [21, 32]] and A@B = [[19, 22], [43, 50]]. The * operator does elementwise multiplication, while @ does standard matrix multiplication.

**Q8.** For x = [10, 20, 30, 40], what happens to ordinary mean if one entry is np.nan? Explain np.nanmean() and why dtype matters.

```python
x = np.array([10.0, 20.0, 30.0, 40.0])
x[2] = np.nan

print(np.mean(x))
print(np.nanmean(x))
```

Ordinary np.mean() returns nan because any operation with NaN produces NaN. np.nanmean() ignores NaN values and averages the rest (23.33). Array dtype matters because NaN requires float arrays (integers cannot hold NaN) and small types can overflow or lose precision.

**Q10.** Predict outputs for a = [10, 20, 30, 40], b = a[1:3], c = a[1:3].copy(), b[0] = 99, c[1] = 77. Explain why b affects a but c does not, and draw a memory diagram.

```python
a = np.array([10, 20, 30, 40])
b = a[1:3]
c = a[1:3].copy()
b[0] = 99
c[1] = 77

print(a)
print(b)
print(c)
```

Outputs: a = [10 99 30 40], b = [99 30], c = [20 77]. b is a slice view pointing to a's memory so modifying b changes a, whereas .copy() allocates separate memory for c.

a -> [10 | 99 | 30 | 40] (b points to middle two)
c -> [20 | 77] (separate block)

**Q11.** For x = [12, 25, 7, 31, 18, 40] and mask = (x > 10) & (x < 30), predict both prints, explain Boolean masking and why & is used instead of and, and rewrite for <= 10 or >= 30.

```python
x = np.array([12, 25, 7, 31, 18, 40])
mask = (x > 10) & (x < 30)
print(mask)
print(x[mask])

m2 = (x <= 10) | (x >= 30)
print(x[m2])
```

Outputs: [True, True, False, False, True, False] and [12, 25, 18]. Boolean indexing picks elements where mask is True. & performs elementwise AND across arrays, while Python's and raises ValueError on multi-element arrays. Rewritten: (x <= 10) | (x >= 30).

**Q14.** For marks matrix M = [[78, 85, 90], [45, 55, 49], [88, 92, 95], [32, 40, 38]], find student and subject averages, students with average < 50, and passing students (>= 40 in all subjects). Explain why array operations help.

```python
M = np.array([
    [78, 85, 90],
    [45, 55, 49],
    [88, 92, 95],
    [32, 40, 38]
])

sa = np.mean(M, axis=1)
sub = np.mean(M, axis=0)
low = np.where(sa < 50)[0]
p = np.where(np.all(M >= 40, axis=1))[0]

print(sa)
print(sub)
print(low)
print(p)
```

Axis operations and boolean masks let us calculate row/column stats and filter conditions across the whole matrix in one step without nested loops.

**Q15.** For T = [28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7], find mean and std dev, mask readings > 40, replace them with the clean mean using np.where(), and explain why anomalies need investigation first.

```python
T = np.array([28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7])
print(np.mean(T), np.std(T))

m = T > 40
avg = np.mean(T[~m])
T2 = np.where(m, avg, T)
print(m)
print(T2)
```

A spike like 65.0 might be a real overheating incident or sensor failure rather than random noise, so auto-replacing it can hide critical alerts.

**Q16.** For image I = [[0, 50, 100], [150, 200, 255], [30, 120, 220]], explain its shape, add 30 and clip to [0, 255], slice top-left 2x2, mask pixels > 150, and explain why arrays suit image/ML tasks.

```python
I = np.array([
    [0, 50, 100],
    [150, 200, 255],
    [30, 120, 220]
])

print(I.shape)
b = np.clip(I + 30, 0, 255)
print(b)
print(I[:2, :2])
print(I > 150)
```

Shape (3, 3) means 3x3 pixels. Representing images as numeric arrays allows fast vectorized cropping, brightness shifts, filtering, and matrix operations for ML models.

**Q17.** For P = [450, 1200, 800, 2500, 1500, 600], apply a 10% discount when purchase >= 1000 using a Boolean mask and np.where(), find total revenue before and after, and explain scaling to thousands of customers.

```python
P = np.array([450, 1200, 800, 2500, 1500, 600])
m = P >= 1000
f = np.where(m, P * 0.9, P)

print(m)
print(f)
print(np.sum(P), np.sum(f))
```

Since comparisons and np.where() are vectorized, the exact same code handles thousands of customers without loops or code changes.

**Q18.** Construct two arrays (5+ numbers) with the same mean but different variances, verify with np.mean() and np.var(), explain variance, and construct two arrays with the same mean and variance but different elements.

```python
a = np.array([8, 9, 10, 11, 12])
b = np.array([0, 5, 10, 15, 20])
print(np.mean(a), np.var(a))
print(np.mean(b), np.var(b))

c = np.array([1, 2, 3, 4, 5])
d = np.array([3 - np.sqrt(5), 3, 3, 3, 3 + np.sqrt(5)])
print(np.mean(c), np.var(c))
print(np.mean(d), np.var(d))
```

Both a and b have mean 10, while variance measures how spread out values are around the mean (2 vs 50). Arrays c and d both have mean 3.0 and variance 2.0 with different values.

**Q19.** Create two arrays of at least 100,000 values, compute $y = 3x^2 + 2x + 1$ using a Python loop and NumPy vectorization, verify equal results, time both across 3 array sizes in a table, and explain vectorization speed and a timing caution.

```python
from time import perf_counter
x1 = np.random.rand(100000)
x2 = np.random.rand(100000)

def exprL(x):
    y = [0.0] * len(x)
    for i in range(len(x)):
        y[i] = 3 * x[i]**2 + 2 * x[i] + 1
    return y
def exprN(x):
    return 3 * x**2 + 2 * x + 1

start = perf_counter()
y1 = exprL(x1)
y2 = exprL(x2)
loop_time = perf_counter() - start

start = perf_counter()
y1n = exprN(x1)
y2n = exprN(x2)
numpy_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")

print(y1 == y1n)
print(y2 == y2n)

a1 = np.random.rand(100)
a2 = np.random.rand(1000)
a3 = np.random.rand(10000)

start = perf_counter()
b1 = exprL(a1)
b2 = exprL(a2)
b3 = exprL(a3)
loop_time = perf_counter() - start

start = perf_counter()
c1 = exprN(a1)
c2 = exprN(a2)
c3 = exprN(a3)
loop_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")
```

| Array Size | Loop Time (s) | NumPy Time (s) |
| :--- | :--- | :--- |
| 100 | ~0.00007 | ~0.00002 |
| 1,000 | ~0.00065 | ~0.00003 |
| 10,000 | ~0.00680 | ~0.00009 |

Vectorization executes compiled C loops over contiguous memory without per-element Python interpreter overhead. Caution: on very small arrays, NumPy call overhead and temporary arrays can be slower than a plain loop, and single timings fluctuate with CPU load.
