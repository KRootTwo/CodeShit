Explain why /, //, and % produce different results.
> '/' in python denotes [division]. As python has [type inference] (determines the datatype automatically). It gives out a float value for thr given input.
> '//' in python denotes [floor division]. Means it gives only the integer value of the division.
> '%' in python denotes [modulus or remainder operator]. As the name suggests it gives the remainder of the division

Explain why the logical expressions produce their respective Boolean values.

> and -> gives True if & only if both coditions are true. Otherwise False.

    [x > y and y > 5] => True and False => [False]

> or -> gives False if & only if both conditions are false. Otherwise True.

    [x > y or y > 5] -> True or False => [True]
 
> not -> also known as negation gives the opposite value.

    [not x > y] -> not( True ) => [False]

```python
#  Modify the values of x and y so that each of the three logical expressions changes its result
x = 10
y = 6
print(x > y and y > 5)
x = 2
y = 3
print(x > y or y > 5)
print(not x > y)
```

3 pairs of numbers (a1, b1), (a1, b2), (a2, b1) for which:
a1 >> b1 == a1 >> b2 == a2 >> b1

a >> b = floor(a / (2**b)):
Let a1 = 3, a2 = 2, b1 = 2, b2 = 3

> (a1, b1) = (3, 2) -> 3 >> 2 = 0
> (a1, b2) = (3, 3) -> 3 >> 3 = 0
> (a2, b1) = (2, 2) -> 2 >> 2 = 0

All 3 evaluate to 0

```python
a1 = 3
b1 = 2
b2 = 3
a2 = 2

print((a1 >> b1) == (a1 >> b2) == (a2 >> b1))
```

```python
# write story.txt
f = open("story.txt", 'w')
f.write('''Python is easy to learn.\nPython is powerful.\nPython is widely used in AI.''')
f.close()

f = open("story.txt", 'r')
print(f.tell())
print(f.read(10))
print(f.tell())
print(f.seek(0))
print(f.readline())
print(f.seek(25))
print(f.read(30))
f.close()
```

> tell() returns the current position of the file pointer.
> seek() moves the file pointer to the specified position in the file.

```python
import re

data = [
    "student@gmail.com",
    "abc@soa.ac.in",
    "9876543210",
    "student123",
    "test@example.com"
]
pattern = r'([\w.-]+)@([a-zA-Z]+).(\w+.?\w+)'

valid_emails = [x for x in data if re.search(pattern, x)]
print("Valid email addresses:")
for e in valid_emails:
    print(e)
```

```python
def student_report(name, roll, branch = 'CSE', **kwargs):
    print("Student Report")
    print(f"Name: {name}")
    print(f"Roll No.: {roll}")
    print(f"Branch: {branch}")
    for i, j in kwargs.items():
        print(f"{i.capitalize()}: {j}")

student_report("Rahul", 101, branch="CSE", cgpa=8.7, city="Bhubaneswar")
```

```python
def data_summary(*args, **kwargs):
    print("DATA SUMMARY")
    print(f"Number of values: {len(args)}")
    print(f"Minimum value: {min(args)}")
    print(f"Maximum value: {max(args)}")
    print(f"Total: {sum(args)}")
    print(f"Average: {sum(args) / len(args)}")
    
    if "dataset" in kwargs:
        print(f"Dataset name: {kwargs['dataset']}")
    if "unit" in kwargs:
        print(f"Unit: {kwargs['unit']}")

data_summary(10, 20, 30, 40, 50, dataset="Marks", unit="percentage")
```

```python
a = 25
b = 40
c = 17

sq = lambda x: x ** 2
mx = lambda x, y: x if x > y else y
ev = lambda x: x % 2 == 0

print(f"square of {a}: {sq(a)}")
print(f"larger of {a} and {b}: {mx(a, b)}")
print(f"is {c} even?: {ev(c)}")
```

> Lambda is useful for quick, throwaway, single-line functions (e.g. used as keys in map(), filter(), or sorted()).
> Standard def functions are better for complex multi-line logic, loops, exception handling, and code that needs maintainability or documentation.

```python
%%writefile calculator.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    if b == 0:
        return "Can't divide by zero"
    return a / b

if __name__ == "__main__":
    print("Testing calculator directly:")
    print("10 + 5 =", add(10, 5))
    print("10 - 5 =", sub(10, 5))
```

```python
import calculator

x = 20
y = 4

print("Addition:", calculator.add(x, y))
print("Subtraction:", calculator.sub(x, y))
print("Multiplication:", calculator.mul(x, y))
print("Division:", calculator.div(x, y))
```

> Thinking Task:
> If `if __name__ == "__main__":` is removed, the test print statements inside `calculator.py` will execute every single time it gets imported into `main.py`.
> The construct ensures code only runs when the script is directly executed by the user, keeping modules clean and reusable as libraries.
