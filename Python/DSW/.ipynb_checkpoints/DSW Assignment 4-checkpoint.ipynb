**Q1.** Write down the variable and its values after execution of each line of the program. Predict all four outputs. Whether changing D will change A? Modify only one line of the program so that changing D does NOT change A.

````python
#Koustav Jana
A = np.arange(1, 7).reshape(2, 3)
B = A[1:, 2].copy()
D = B.T
D[0] = 100

print(A)
print(B)
print(D)
````

A is [[1, 2, 3], [4, 5, 6]] and B is [6] (view of A[1, 2]). Line 3 (C = A[1:][2]) gives IndexError since A[1:] has only 1 row. Yes, changing D changes A because B and D are views sharing memory with A. Changing line 2 to B = A[1:, 2].copy() prevents this.

**Q3.** Given a matrix representing marks of a student in different subjects(columns), and different semesters(Rows) identify where the mark is 50+ in at least two subjects. Write the algorithm. Draw a flowchart. Take your semester 1, semester 2, and expected semester 3 marks, write the code, and execute this program.

````python
#Koustav Jana
m = np.array([
    [76, 81, 44, 85, 72],
    [47, 83, 41, 78, 80],
    [84, 87, 79, 89, 82]
])

c = np.sum(m >= 50, axis=1)
r = np.where(c >= 2)[0] + 1
print(c)
print(r)
````

3. Find rows where count >= 2 and print semester numbers.

Flowchart: Start -> Input m -> c = sum(m >= 50, axis=1) -> Find c >= 2 -> Print semesters -> End

**Q4.** Consider an array containing the integers from 1 to 24. Write NumPy code to create this array. Determine which of the following shapes are valid for reshaping the array: (2,12), (3,8), (4,6), (2,3,4), (5,5). Explain mathematically why some shapes are valid, and others are not.

**Q5.** Suppose A.shape(4, 7, 1). For each of the following shapes, state whether NumPy broadcasting is possible when combining A elementwise with an array of that shape: (7,1), (4,7), (1,7,1), (4, 1,5), (3, 7, 1). If broadcasting is possible, give the shape of the output.

````python
#Koustav Jana
a = np.ones((4, 7, 1))
for s in [(7, 1), (4, 7), (1, 7, 1), (4, 1, 5), (3, 7, 1)]:
    try:
        b = a + np.ones(s)
        print(s, b.shape)
    except ValueError:
        print(s, "Not possible")
````

Matching trailing dimensions with (4, 7, 1): (7,1) -> (4,7,1), (4,7) -> Not possible (7 != 1), (1,7,1) -> (4,7,1), (4,1,5) -> (4,7,5), (3,7,1) -> Not possible (3 != 4).

**Q8.** A data analyst has the following array: x = [10, 20, 30, 40]. What happens to the ordinary mean if one entry is replaced by np.nan? Explain the purpose of np.nanmean() and why the data type of an array can matter during numerical computation.

````python
#Koustav Jana
x = np.array([10.0, 20.0, 30.0, 40.0])
x[2] = np.nan

print(np.mean(x))
print(np.nanmean(x))
````

Ordinary np.mean() returns nan because any operation with NaN produces NaN. np.nanmean() ignores NaN values and averages the rest (23.33). Array dtype matters because NaN requires float arrays (integers cannot hold NaN) and small types can overflow or lose precision.

**Q11.** Consider x = np.array([12, 25, 7, 31, 18, 40]), mask = (x > 10) & (x < 30). Predict the output of both print statements. Explain how the Boolean mask selects elements. Why is & used here instead of the Python keyword and? Rewrite the condition to select values less than or equal to 10, or greater than or equal to 30.

````python
#Koustav Jana
x = np.array([12, 25, 7, 31, 18, 40])
mask = (x > 10) & (x < 30)
print(mask)
print(x[mask])

m2 = (x <= 10) | (x >= 30)
print(x[m2])
````

Outputs: [True, True, False, False, True, False] and [12, 25, 18]. Boolean indexing picks elements where mask is True. & performs elementwise AND across arrays, while Python's and raises ValueError on multi-element arrays. Rewritten: (x <= 10) | (x >= 30).

**Q14.** A teacher records marks for four students in three subjects: $M=\begin{bmatrix}78&85&90\\ 45&55&49\\ 88&92&95\\ 32&40&38\end{bmatrix}$. Rows represent students and columns represent subjects. Write NumPy code to create M. Calculate the average mark of each student and the average mark in each subject. Use Boolean indexing to identify students whose average is below 50. Define a student as passing only if the student scores at least 40 in every subject. Use NumPy to identify the students who pass. Explain how array operations make this analysis more convenient than checking each value individually.

````python
#Koustav Jana
M = np.array([
    [78, 85, 90],
    [45, 55, 49],
    [88, 92, 95],
    [32, 40, 38]
])

sa = np.mean(M, axis=1)
sub = np.mean(M, axis=0)
low = np.where(sa < 50)[0]
p = np.where(np.all(M >= 40, axis=1))[0]

print(sa)
print(sub)
print(low)
print(p)
````

Axis operations and boolean masks let us calculate row/column stats and filter conditions across the whole matrix in one step without nested loops.

**Q15.** A temperature sensor records the following readings in degrees Celsius: T = [28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7]. Calculate the mean and standard deviation using NumPy. Create a Boolean mask to identify readings greater than $40^{\circ}C$. Use np.where() to replace readings above $40^{\circ}C$ with the mean of the non-anomalous readings. Explain why a sensor anomaly should be investigated before automatically replacing the reading.

````python
#Koustav Jana
T = np.array([28.5, 29.1, 28.9, 29.4, 65.0, 29.2, 28.7])
print(np.mean(T), np.std(T))

m = T > 40
avg = np.mean(T[~m])
T2 = np.where(m, avg, T)
print(m)
print(T2)
````

A spike like 65.0 might be a real overheating incident or sensor failure rather than random noise, so auto-replacing it can hide critical alerts.

Explain the meaning of the array shape in this example. Use NumPy to increase every pixel intensity by 30, then clip values to the range 0 to 255. **Q17.** A shop stores the total purchase amount for six customers: P = [450, 1200, 800, 2500, 1500, 600]. The shop offers a 10% discount to customers whose purchase is at least 1000. Create a Boolean mask identifying eligible customers. Use np.where() to calculate the final amount for every customer. Calculate the total revenue before and after discounts. Explain how the same vectorized rule could be applied to thousands of customers.

````python
#Koustav Jana
P = np.array([450, 1200, 800, 2500, 1500, 600])
m = P >= 1000
f = np.where(m, P * 0.9, P)

print(m)
print(f)
print(np.sum(P), np.sum(f))
````

Since comparisons and np.where() are vectorized, the exact same code handles thousands of customers without loops or code changes.

**Q18.** Construct two NumPy arrays, each containing at least five numbers, that have the same mean but different variances. Write down both arrays and calculate their means by hand. Use np.mean() and np.var() to verify your results. Explain what the variance tells us that the mean alone does not. Next, try to construct two arrays with the same mean and the same variance but different element values. Verify your example.

````python
#Koustav Jana
a = np.array([8, 9, 10, 11, 12])
b = np.array([0, 5, 10, 15, 20])
print(np.mean(a), np.var(a))
print(np.mean(b), np.var(b))

c = np.array([1, 2, 3, 4, 5])
d = np.array([3 - np.sqrt(5), 3, 3, 3, 3 + np.sqrt(5)])
print(np.mean(c), np.var(c))
print(np.mean(d), np.var(d))
````

Both a and b have mean 10, while variance measures how spread out values are around the mean (2 vs 50). Arrays c and d both have mean 3.0 and variance 2.0 with different values.

**Q19.** Create two arrays, each containing at least 100,000 numerical values. Compute the expression $y=3x^{2}+2x+1$ using (i) a Python loop and (ii) NumPy vectorized operations. Write both implementations. Check whether the two methods produce the same numerical results. Measure and record the execution time of each method using Python's time module. Repeat the experiment at three different array sizes and present your observations in a table. Explain why vectorization can be faster. Mention at least one limitation or caution when comparing execution times.

````python
#Koustav Jana
from time import perf_counter
x1 = np.random.rand(100000)
x2 = np.random.rand(100000)

def exprL(x):
    y = [0.0] * len(x)
    for i in range(len(x)):
        y[i] = 3 * x[i]**2 + 2 * x[i] + 1
    return y
def exprN(x):
    return 3 * x**2 + 2 * x + 1

start = perf_counter()
y1 = exprL(x1)
y2 = exprL(x2)
loop_time = perf_counter() - start

start = perf_counter()
y1n = exprN(x1)
y2n = exprN(x2)
numpy_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")

print(y1 == y1n)
print(y2 == y2n)

a1 = np.random.rand(100)
a2 = np.random.rand(1000)
a3 = np.random.rand(10000)

start = perf_counter()
b1 = exprL(a1)
b2 = exprL(a2)
b3 = exprL(a3)
loop_time = perf_counter() - start

start = perf_counter()
c1 = exprN(a1)
c2 = exprN(a2)
c3 = exprN(a3)
loop_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")
````

```text
Array Size | Loop Time (s) | NumPy Time (s)
100        |  0.00007      |  0.00002
1,000      |  0.00065      |  0.00003
10,000     |  0.00680      |  0.00009
```

Vectorization executes compiled C loops over contiguous memory without per-element Python interpreter overhead. Caution: on very small arrays, NumPy call overhead and temporary arrays can be slower than a plain loop, and single timings fluctuate with CPU load.
