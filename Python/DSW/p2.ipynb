```python
# Koustav Jana
l = [4, 2, 4, 3, 2, 3, 3, 2, 5, 3, 3]
d = dict()
for i in l:
    if i not in d.keys():
        d[i] = 1
    else :
        d[i] += 1
print(d)

most = -1
most2 = -1
for i, j in d.items():
    if j > most:
        most2 = most
        most = j
vFk = lambda x: [i for i, j in d.items() if j == x]

print(f"2nd most freq element: {vFk(most2)}")
```

```python
# Koustav Jana
l = [8, 3, 12, 20, 7, 15]
p = []
for i in l:
    if i-1 not in l:
        q = []
        while i in l:
            q.append(i)
            i += 1
        if len(q) == 2:
            p.append(q)
print(p)
```

```python
# Koustav Jana
l = [1, 2, 3 , 9, 0]
max_list = []
curr_list = []
for i in l:
    if i-1 not in l:
        curr_list = []
        while i in l:
            curr_list.append(i)
            i = i + 1
    if len(curr_list) > len(max_list) :
        max_list = curr_list
    
print(max_list)
```

```python
# Koustav Jana
l = [2, 7, 11, 15, 3, 6]
key = 9
p = []
for i in range(len(l) - 1) :
    for j in range(i + 1, len(l)) :
        if (l[i] + l[j]) == key:
            p.append((l[i], l[j]))
print(p)
```
