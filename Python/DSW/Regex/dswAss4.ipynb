### Question 19
Create two arrays, each containing at least 100,000 numerical values. Compute the expression:
$$y = 3x^2 + 2x + 1$$
using (i) a Python loop and (ii) NumPy vectorized operations.
- Write both implementations.
- Check whether the two methods produce the same numerical results.
- Measure and record the execution time of each method using Python's `time` module.
- Repeat the experiment at three different array sizes and present your observations in a table.
- Explain why vectorization can be faster. Mention at least one limitation or caution when comparing execution times.

````python
from time import perf_counter
x1 = np.random.rand(100000)
x2 = np.random.rand(100000)

def exprL(x):
    y = [0.0] * len(x)
    for i in range(len(x)):
        y[i] = 3 * x[i]**2 + 2 * x[i] + 1
    return y
def exprN(x):
    return 3 * x**2 + 2 * x + 1

start = perf_counter()
y1 = exprL(x1)
y2 = exprL(x2)
loop_time = perf_counter() - start

start = perf_counter()
y1n = exprN(x1)
y2n = exprN(x2)
numpy_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")

print(y1 == y1n)
print(y2 == y2n)

a1 = np.random.rand(100)
a2 = np.random.rand(1000)
a3 = np.random.rand(10000)

start = perf_counter()
b1 = exprL(a1)
b2 = exprL(a2)
b3 = exprL(a3)
loop_time = perf_counter() - start

start = perf_counter()
c1 = exprN(a1)
c2 = exprN(a2)
c3 = exprN(a3)
loop_time = perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy vectorized: {numpy_time:.6f} seconds")
````
