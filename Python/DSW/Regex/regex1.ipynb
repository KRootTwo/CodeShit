```python
# Koustav Jana
def find_course_codes(text):
    pattern = r'\b[A-Z]{2}\d{3}\b'
    course_codes = re.findall(pattern, text)
    return course_codes
text = '''Here are some course codes: CS101, MA202, PH303.
        Note that CS202 & AB100 are valid, but not CS1010 or MA20.'''
codes = find_course_codes(text)
print("course codes found: ", codes)
```

```python
# Koustav Jana
def find_email_ids(text):
    pattern = r'[a-zA-Z0-9.%+=]+@[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}'
    email_ids = re.findall(pattern, text)
    return email_ids
text = '''Here are some email addresses:
- avani-lekhara@soa.ac.in
- nitesh.kumar@gmail.com
- sumit_antil@yahoo.com
- HarvinderSingh@domain.org
- Dharabir@domain
- Praveen+Kumar@rediff.mail
- Navdeep@Singh.in'''
emails = find_email_ids(text)
print("Email address found: ", emails)
```
